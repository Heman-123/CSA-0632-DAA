```python
import os
import time
from datetime import datetime

# File name
filename = "evidence.txt"

# Function to display file timestamps
def show_metadata(file):
    stats = os.stat(file)

    print("----- File Metadata -----")
    print("Created Time     :", datetime.fromtimestamp(stats.st_ctime))
    print("Modified Time    :", datetime.fromtimestamp(stats.st_mtime))
    print("Last Access Time :", datetime.fromtimestamp(stats.st_atime))
    print()

# Step 1: Create a file
with open(filename, "w") as file:
    file.write("Initial forensic evidence.\n")

print("File created successfully.\n")

# Display metadata after creation
print("Metadata After File Creation:")
show_metadata(filename)

# Wait for a few seconds
time.sleep(5)

# Step 2: Modify the file
with open(filename, "a") as file:
    file.write("Additional evidence added.\n")

print("File modified successfully.\n")

# Display metadata after modification
print("Metadata After File Modification:")
show_metadata(filename)

print("Locard's Exchange Principle Demonstrated:")
print("Any interaction with the file leaves traces in its metadata.")
```
