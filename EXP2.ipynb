```python
import hashlib

# Create a dummy evidence.txt file for demonstration if it doesn't exist
try:
    with open("evidence.txt", "x") as f:
        f.write("This is some sample evidence data.\n")
        f.write("Line two of evidence.\n")
except FileExistsError:
    pass # File already exists, proceed

# Function to calculate SHA-256 hash
def calculate_hash(filename):
    sha256 = hashlib.sha256()
    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)
    return sha256.hexdigest()

# File to verify
filename = "evidence.txt"

# Step 1: Calculate original hash
original_hash = calculate_hash(filename)

print("Original SHA-256 Hash:")
print(original_hash)

# Step 2: Verify file integrity
current_hash = calculate_hash(filename)

print("\nCurrent SHA-256 Hash:")
print(current_hash)

# Step 3: Compare hashes
if original_hash == current_hash:
    print("\nVerification Successful!")
    print("Chain of Custody Maintained.")
    print("The digital evidence has not been modified.")
else:
    print("\nVerification Failed!")
    print("Chain of Custody Broken.")
    print("The digital evidence has been altered.")
```
